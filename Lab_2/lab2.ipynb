# Lab2:  Generation of Random Number Using Pseudo Random and Chi-Square Test

## Aim

To generate pseudo-random numbers using a linear congruential generator (LCG) and test the sequence for uniformity using the chi-square goodness-of-fit test.

## Objective

To study the working of a pseudo-random number generator and verify whether the generated numbers are uniformly distributed.

## Theory

A linear congruential generator is a simple method for generating pseudo-random numbers using the recurrence relation $r_{i+1} = (a r_i + b) \bmod p$.

In this experiment, the generated values are divided into equal bins and the chi-square test is applied. If the calculated chi-square value is less than the critical value, the sequence is considered to follow a uniform distribution.

## Application

Pseudo-random number generation is widely used in simulation, statistical sampling, Monte Carlo methods, numerical analysis, gaming, and computer-based modeling.

## Algorithm

1. Select the LCG parameters: multiplier, increment, modulus, and initial seed.
2. Generate the pseudo-random sequence using the LCG formula.
3. Divide the values into equal bins.
4. Count the number of values in each bin.
5. Compute the chi-square statistic from the observed and expected frequencies.
6. Compare the calculated value with the critical value at 5% significance level.
7. Conclude whether the generated sequence passes the uniformity test.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2
# Constants
iterator = 256 # Number of random numbers to generate
a = 29 # Multiplier
b = 37 # Increment
p = 256 # Modulus
n = 8 # Number of bins for Chi-Square test
x02 = 0 # Chi-Square statistic initialization
critical_value = chi2.ppf(1 - 0.05, df = n - 1) # Chi-Square critical value
# Initialize arrays
r = np.ones(shape=(iterator), dtype=int) # Array to store generated random numbers
ei = iterator / n # Expected frequency for each bin
c = np.zeros(shape=(n), dtype=int) # Bin counts
# Initial seed
r[0] = 17
# Generate random numbers using LCG
for i in range(0, iterator - 1):
    r[i + 1] = (r[i] * a + b) % p
# Bin the generated numbers and compute frequencies
for i in range(iterator):
    if r[i] <= 31:
        c[0] += 1
    elif r[i] <= 63:
        c[1] += 1
    elif r[i] <= 95:
        c[2] += 1
    elif r[i] <= 127:
        c[3] += 1
    elif r[i] <= 159:
        c[4] += 1
    elif r[i] <= 191:
        c[5] += 1
    elif r[i] <= 223:
        c[6] += 1
    elif r[i] <= 255:
        c[7] += 1
# Compute Chi-Square statistic
for i in range(n):
    x02 += ((c[i] - ei) ** 2) / ei
# Print results
print("Values of r:", r)
print("Values of c (bin frequencies):", c)
print("Value of Chi-Square statistic (x02):", x02)
print("The critical value is", critical_value)
print("Chi square test passed" if x02 <= critical_value else "Chi squsre test failed")
# Visualize the generated random numbers
plt.figure()
plt.plot(r, label="Generated Sequence (r)")
plt.xlabel("Index")
plt.ylabel("Random Number")
plt.title("LCG Generated Sequence")
plt.legend()
plt.show()
```

## Result / Observation

The generated values were distributed equally across all 8 bins, with each bin containing 32 values. The chi-square statistic obtained was 0.0, which is less than the critical value at the 5% significance level. Therefore, the generated sequence passed the chi-square test for uniformity.

## Discussion

The linear congruential generator produced 256 pseudo-random values using the given multiplier, increment, modulus, and initial seed. These values were grouped into 8 bins of equal size, and the expected frequency for each bin was 32.

From the output, all bins contained exactly 32 values, so the chi-square statistic came out to be 0.0. Since this value is much smaller than the critical value, the sequence satisfies the uniformity test for this experiment.

This result shows that the chosen LCG parameters generated a well-balanced sequence for the selected sample. However, one test alone is not enough to prove that a sequence is completely random. More statistical tests would be needed for a deeper evaluation.

## Conclusion

In this experiment, pseudo-random numbers were generated using the linear congruential method and checked using the chi-square test. The generated sequence passed the test because the observed frequencies matched the expected frequencies exactly. Hence, for this set of parameters and seed, the sequence can be considered uniformly distributed for the given sample size.
